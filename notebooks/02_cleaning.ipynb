Загрузка данных и инициализация

```python
import sys
from pathlib import Path

sys.path.append(str(Path("../src").resolve()))
from cleaning import *

df = pd.read_parquet("../data/interim/games_typed.parquet")
display(
    f"Загружено строк: {len(df)}, колонок: {df.shape[1]}"
)  
```

Этап 1: Приведение имён столбцов

```python
rows_before = len(df)
df = normalize_column_names(df)
display(f"Строк до: {rows_before}, после: {len(df)}")
```

Этап 2: Обработка пропущенных значений

```python
rows_before = len(df)
df = handle_missing_values(df)
display(f"Строк до: {rows_before}, после: {len(df)}")
```

Этап 3: Удаление дубликатов

```python
rows_before = len(df)
df = remove_duplicates(df)
display(f"Строк до: {rows_before}, после: {len(df)}")
```

Этап 4: Нормализация категорий

```python
rows_before = len(df)
df = normalize_categories(df)
display(f"Строк до: {rows_before}, после: {len(df)}")
```

Этап 5: Валидация и удаление выбросов

```python
rows_before = len(df)
df = validate_and_filter_outliers(df)
display(
          f"Строк до: {rows_before}, после: {len(df)}. Итоговый чистый набор готов!"
  )
```
